```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib_fontja
```

```python
def f(x):
    return x

def g(x):
    return np.log(x)


def fg(x):
    return g(x)*f(x)

x = np.linspace(1e-8, 1.5, 100)
y = fg(x)

plt.plot(x, y)
plt.grid(alpha=0.2)
plt.show()
```
